## Step 12: Feature Extraction (Month & Year from Date)

### Observation:
1. "Month and Year columns were successfully created from the Date column. These new features will help in performing time-based sales analysis and visualization."


## Step 13: Feature Engineering

```python
df.columns
```

### Observation:
"The Month and Year columns were successfully added to the dataset. These new features will be used for time-based sales analysis and visualization."

## Step 13A: Day Name

```python
df["Day Name"] = df["Date"].dt.day_name()
```

## Step 13B: Quarter

```python
df["Quarter"] = df["Date"].dt.quarter
```

## Step 13C: Weekend or Weekday

```python
df["Weekend/Weekday"] = df["Date"].dt.dayofweek.apply(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)
```

## Step 13D: Customer Segment (based on spending)

```python
df["Customer Segment"] = pd.cut(
    df["Total Amount"],
    bins=[0, 100, 500, 1000, df["Total Amount"].max()],
    labels=["Low", "Medium", "High", "Premium"]
)
```

## Step 13E: Revenue Category (based on Total Amount)

```python
df['Revenue Category'] = pd.cut(
    df['Total Amount'],
    bins=[0, 100, 300, df['Total Amount'].max()],
    labels=['Low', 'Medium', 'High']
)

# Check the result
df[['Total Amount', 'Revenue Category']].head()
```

### Feature engineering was performed to create additional analytical variables such as Day Name, Quarter, Weekend/Weekday, Customer Segment, and Revenue Category, enabling deeper sales and customer behavior analysis.

## Step 16: Revenue Analysis by Gender

### Purpose
To compare the total revenue generated by male and female customers and understand which customer group contributes more to overall sales.


```python
gender_sales = df.groupby("Gender")["Total Amount"].sum()

gender_sales
```

### Visualization

```python
# Revenue by gender
revenue_by_gender = df.groupby('Gender')['Total Amount'].sum().sort_values()

plt.figure(figsize=(7,4.5))
bars = plt.barh(revenue_by_gender.index, revenue_by_gender.values)

plt.title('Revenue by Gender', fontsize=14, fontweight='bold')
plt.xlabel('Total Revenue (₹)')
plt.ylabel('Gender')

plt.grid(axis='x', alpha=0.3)

# Value labels
for bar in bars:
    plt.text(bar.get_width() + 1000,
             bar.get_y() + bar.get_height()/2,
             f'₹{bar.get_width():,.0f}',
             va='center', fontsize=10)

plt.tight_layout()
plt.show()
```

### Observation
1. The chart compares the total revenue generated by male and female customers.

2. One gender contributes slightly higher revenue than the other, but the difference is not extremely large.

3. Revenue generation is relatively balanced across both customer groups.

### Business Insight
1. Continue targeting both male and female customers equally.

2. Design personalized marketing campaigns based on purchasing patterns instead of gender alone.

3. Introduce loyalty programs to encourage repeat purchases.

## Step 17: Customer Age Group Segmentation

### Purpose
To divide customers into different age groups for better analysis.

```python
age_bins = [18, 25, 35, 45, 55, 65]

age_labels = ["18-25", "26-35", "36-45", "46-55", "56-64"]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

df[["Age", "Age Group"]].head()
```

### Observation
1. Revenue changes across different age groups, with some age segments contributing significantly more than others.

2. The line chart highlights the trend in customer spending across age categories.

3. Middle-age customer groups generally show stronger purchasing activity than younger or older segments.

### Business Insights
1. Marketing campaigns should focus on the age groups generating the highest revenue.

2. Product assortments can be optimized according to the preferences of high-value age segments.

3. Age-based customer segmentation can improve advertising effectiveness and customer retention.

## Step 18: Revenue Analysis by Age Group

### Purpose
To identify which age group contributes the highest revenue and understand the spending pattern of different age categories.

```python
age_group_sales = df.groupby("Age Group")["Total Amount"].sum()

age_group_sales
```

### Visualization

```python
# Revenue by age group
age_revenue = df.groupby('Age Group')['Total Amount'].sum().reindex(
    ['18-25', '26-35', '36-45', '46-55', '56-64']
)

plt.figure(figsize=(8,5))

plt.fill_between(age_revenue.index, age_revenue.values, alpha=0.4)
plt.plot(age_revenue.index, age_revenue.values, marker='o', linewidth=2)

plt.title('Revenue by Age Group', fontsize=14, fontweight='bold')
plt.xlabel('Age Group')
plt.ylabel('Total Revenue (₹)')

plt.grid(axis='y', alpha=0.3)

# Value labels
for x, y in zip(age_revenue.index, age_revenue.values):
    plt.text(x, y + 1000, f'₹{y:,.0f}', ha='center', fontsize=9)

plt.tight_layout()
plt.show()
```

1. The 46–55 age group generates the highest total revenue among all age categories.

2. Customers aged 26–45 also contribute a substantial share of revenue, indicating strong purchasing activity in the middle-age segments.

3. The 18–25 age group contributes comparatively lower revenue, suggesting lower average spending or fewer high-value purchases.

### Business Insights
1.  Develop premium products and exclusive offers for high-spending age groups.

2. Launch targeted marketing campaigns to improve engagement among younger customers.

3. Personalize promotions based on customer age segments.


## Step 19: Monthly Revenue Analysis

### Purpose
To analyze the total sales for each month and identify the months with the highest and lowest sales.

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

monthly_sales = (
    df.groupby("Month")["Total Amount"]
      .sum()
      .reindex(month_order)
)

monthly_sales
```

### Visualization

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=3
)

plt.title("Monthly Revenue Trend", fontsize=16, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Revenue (₹)", fontsize=12)
for i, value in enumerate(monthly_sales.values):
    plt.text(
        i,
        value + 500,
        f"₹{value:,.0f}",
        ha="center",
        fontsize=9,
        fontweight="bold"
    )
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### Observation
1. Monthly revenue fluctuates throughout the year.

2. Some months recorded significantly higher sales than others.

3. These variations may be influenced by seasonal demand, promotional campaigns, or customer purchasing behavior.

### Business Insight
1. Identify the reasons behind high-performing months and replicate successful strategies.

2. Introduce promotional campaigns during low-sales months.

3. Use monthly sales trends to improve inventory planning and business forecasting.

## Step 21: Average Customer Spending by Product Category

To understand the distribution of low, medium, and high revenue transactions for male and female customers and compare their contribution across revenue categories.

### Visualization

```python
# Revenue category by gender
revenue_category_gender = pd.crosstab(df['Revenue Category'], df['Gender'])

# Arrange categories in order
revenue_category_gender = revenue_category_gender.reindex(['Low', 'Medium', 'High'])

plt.figure(figsize=(7,5))

revenue_category_gender.plot(
    kind='bar',
    stacked=True,
    figsize=(7,5)
)

plt.title('Revenue Category Distribution by Gender', fontsize=14, fontweight='bold')
plt.xlabel('Revenue Category')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.legend(title='Gender')

plt.tight_layout()
plt.show()
```

### Observation

1. The chart shows the distribution of low, medium, and high revenue transactions for both male and female customers.

2. Medium revenue transactions represent a significant portion of total transactions.

3. Both genders contribute across all revenue categories, while high-value transactions are comparatively fewer.
